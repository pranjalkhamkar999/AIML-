```python
import math
import numpy as np
from scipy import stats
```

```python
# Math Library
num = 16
print(f"Square root of {num}: {math.sqrt(num)}")
print(f"Factorial of 5: {math.factorial(5)}")
print(f"log2(8): {math.log2(8)}")
```

```python
# NumPy Library
arr1 = np.array([1, 2, 3, 4, 5])
arr2 = np.array([10, 20, 30, 40, 50])

print(f"Array Addition:       {arr1 + arr2}")
print(f"Array Dot Product:    {np.dot(arr1, arr2)}")
print(f"Reshaped 2x3 Matrix:\n{np.arange(6).reshape(2, 3)}")
```

```python
# SciPy Library
sample = [20, 22, 23, 20, 25, 27, 26, 20, 29]
scipy_mode = stats.mode(sample, keepdims=True)
scipy_zscore = stats.zscore(sample)

print(f"Mode: {scipy_mode.mode[0]} (Count: {scipy_mode.count[0]})")
print(f"Z-scores: {np.round(scipy_zscore, 2)}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Pandas: Data Handling
data = {
    'Student': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank'],
    'Study_Hours': [2.5, 5.1, 3.2, 8.5, 3.5, 6.0],
    'Score': [54, 76, 60, 95, 65, 82]
}

df = pd.DataFrame(data)
print("DataFrame")
print(df)
print("\nSummary Statistics:")
print(df.describe())
```

```python
# Matplotlib & Seaborn: Visualization
plt.figure(figsize=(12, 5))

# Subplot 1: Matplotlib Line & Scatter Plot
plt.subplot(1, 2, 1)
plt.scatter(df['Study_Hours'], df['Score'], color='crimson', label='Scores')
plt.plot(df['Study_Hours'], df['Score'], color='gray', linestyle='--', alpha=0.5)
plt.title('Study Hours vs Score (Matplotlib)')
plt.xlabel('Study Hours')
plt.ylabel('Score')
plt.grid(True)
plt.legend()

# Subplot 2: Seaborn Regression Plot
plt.subplot(1, 2, 2)
sns.regplot(x='Study_Hours', y='Score', data=df, color='royalblue')
plt.title('Linear Fit (Seaborn)')
plt.xlabel('Study Hours')
plt.ylabel('Score')

plt.tight_layout()
plt.show()
```
